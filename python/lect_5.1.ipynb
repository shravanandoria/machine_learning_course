Lambda Functions

A lambda function in Python is a small, anonymous function that is defined without a name. Unlike standard functions created using the def keyword, lambda functions are written in a single line, can take any number of arguments, but can only contain a single expression

```python
def addition(a, b):
    return a + b

print(addition(8,2))
```

```python
add_num=lambda a,b:a+b

print(add_num(8,2))
print(add_num(10,5))
print(add_num(20,8))
```

```python
multiply__num= lambda a,b:a*b

print(multiply__num(8,2))
```

```python
check_even=lambda num:num % 2 == 0

print(check_even(4))
print(check_even(5))
```
