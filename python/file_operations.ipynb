File Operation - Read & Write Files

File handling in Python allows you to create, read, update, and delete files stored on your computer

```python
# Read a file
with open('example.txt', mode='r') as file:
    content = file.read()
    print(content)
```

```python
# Read a file line by line

with open('example.txt',mode='r') as file:
    for line in file.readlines():
        print(line.strip())
```

```python
# Write a file(Overwriting)

with open('shravan.txt', mode='w') as file:
    file.write("I'm shravan\n")
    file.write('I love travelling\n')
    file.write('I listen to music\n')
```

```python
with open('shravan.txt', mode='w') as file:
    file.write('i love playing video game')
```

```python
# Write a file (without overwriting)

with open('shravan.txt', mode='a') as file:
    file.write('\ni love playing video game')
```

```python
# Writing a list of lines to a file

my_text = ['playing video games\n', 'listening music\n', 'travelling\n']

with open('my_hobies.txt', mode='w') as file:
    file.writelines(my_text)
```

```python
# Binary Files

binary_text=b'\x00\x01\x02\x03\x04'

with open('binary_text.txt', 'wb') as file:
    file.write(binary_text)
```

```python
# Reading a binary file

with open('binary_text.txt', 'rb') as file:
    content = file.read()
    print(content)
```

```python
# Read the content from source file & write to destination file

with open('source.txt', 'r') as file:
    content = file.read()
    
with open('destination.txt', 'w') as file:
    file.write(content)
```

```python
# Read a text file & count lines, words, characters in it

def count_data(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        lines_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
        
        return lines_count, word_count, char_count
    
file_path = 'source.txt'

lines_count, word_count, char_count = count_data(file_path)

print(lines_count, word_count, char_count)
```

```python
# writing & then reading a file

with open('superheros.txt', 'w+') as file:
    file.write('superman is strong\n')
    file.write('Iron man is smart\n')
    file.write('spiderman is fast\n')
    
    file.seek(0)
    
    content = file.read()
    print(content)
```
