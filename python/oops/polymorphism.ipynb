# Polymorphism

Polymorphism in Python means "many forms". It refers to the ability of different object types to respond to the same function name, method, or operator in their own unique way, it is typically achieved through method overriding & interface

# Method Overriding

Method overriding in Python occurs when a child class defines a method with the same name, return type, and parameters as a method already existing in its parent class

```python
class Animal:
    def __init__(self):
        pass
    
    
    def make_sound(self):
        """Create your own implementation of this method"""
        pass
```

```python
class Dog(Animal):
    def __init__(self, name):
        self.name = name
        
    def make_sound(self):
        return "the dog says woof!!!"
```

```python
class Cat(Animal):
    def __init__(self, name):
        self.name = name
        
    def make_sound(self):
        return "the cat says meow!!!"
```

```python
dog=Dog('shadow')
cat=Cat('kitty')

def sounds_of_animal(animal_obj):
    print(animal_obj.make_sound())
    
    
sounds_of_animal(dog)
sounds_of_animal(cat)
```

```python
class Shape:
    def __init__(self):
        pass
    
    def area(self):
        """create your own implementation of calculating area"""
```

```python
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
        
    def area(self):
        return self.width * self.height
```

```python
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
        
    def area(self):
        return 3.14 * self.radius * self.radius
```

```python
rectangle=Rectangle(10, 20)
circle=Circle(5)



def print_area(shape_obj):
    print(shape_obj.area())

print_area(rectangle)
print_area(circle)
```

# Abstract Base Class

In Python, an Abstract Base Class (ABC) is a class that cannot be instantiated directly and serves as a formal blueprint for other classes

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    def __init__(self):
        pass
    
    @abstractmethod
    def start_engine(self):
        """create your own implementation of this method"""
        pass
```

```python
class Car(Vehicle):
    def __init__(self):
        pass
    
    def start_engine(self):
        return "car engine is started"
```

```python
class Bike(Vehicle):
    def __init__(self):
        pass
    
    def start_engine(self):
        return "bike engine is started"
```

```python
# vehicle = Vehicle()
car = Car()
bike = Bike()

print(car.start_engine())
print(bike.start_engine())
```
