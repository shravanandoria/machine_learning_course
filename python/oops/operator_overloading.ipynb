# Operator Overloading

Operator overloading in Python allows you to redefine how built-in operators (like +, -, *, ==, etc.) behave when used with custom objects

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, other):
        return Vector(self.x * other.x, self.y * other.y)

    def __str__(self):
        return f"{self.x}: {self.y}"
```

```python
vec1 = Vector(2,3)
vec2 = Vector(5,8)

print(vec1 + vec2)
print(vec1 - vec2)
print(vec1 * vec2)
```
