# Encapsulation

Encapsulation is a core concept of Object-Oriented Programming that involves bundling data and the methods that operate on that data into a single unit

```python
class Person:
    def __init__(self, name, age):
        self.name = name # Public attribute
        self.age = age # Public attribute
        

person = Person('shravan', 27)

print(person.name)
print(person.age)
        
```

```python
class Person:
    def __init__(self, name, age, hair_color):
        self.__name = name # Private attribute
        self.__age = age # Private attribute
        self.hair_color = hair_color # Public attribute
        
    def get_name(self):
        return self.__name
    
    def get_age(self):
        return self.__age
```

```python
class Employee(Person):
    def __init__(self, name, age, hair_color, salary):
        super().__init__(name, age, hair_color)
        self.salary = salary
        
```

```python
class Person:
    def __init__(self, name, age, hair_color):
        self.__name = name # Private attribute
        self.__age = age # Private attribute
        self.hair_color = hair_color # Public attribute
        
    def get_name(self):
        return self.__name
    
    def get_age(self):
        return self.__age
    
    def set_name(self, new_name):
        self.__name = new_name
        
    def set_age(self, new_age):
        self.__age = new_age
```

```python
person = Person('alex', 28, 'brown')
print(person.get_name())
print(person.get_age())

person.set_name('bob')
person.set_age(40)

print(person.get_name())
print(person.get_age())
```

```python
class Person:
    def __init__(self, name, age, hair_color):
        self._name = name # Protected attribute
        self._age = age # Protected attribute
        self.hair_color = hair_color # Public attribute
        
    def get_name(self):
        return self.__name
    
    def get_age(self):
        return self.__age
    
    def set_name(self, new_name):
        self.__name = new_name
        
    def set_age(self, new_age):
        self.__age = new_age
```

```python
class Employee(Person):
    def __init__(self, name, age, hair_color, salary):
        super().__init__(name, age, hair_color)
        self.salary = salary
        
```

```python
employee = Employee('shravan', 27, 'black', 2000)

print(employee._name)
print(employee._age)
```
