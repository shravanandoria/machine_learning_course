# Inheritance

```python
class Vehicle:
    def __init__(self, name, color, engine_type):
        self.name = name
        self.color = color
        self.engine_type = engine_type
    
    def accelerate(self):
        print(f"{self.name} is accelerating")
        
    def brake(self):
        print(f"{self.name} applies brake")
```

```python
class Car(Vehicle):
    def __init__(self, name, color, engine_type, is_self_driving, gear_type):
        super().__init__(name, color, engine_type)
        self.is_self_driving = is_self_driving
        self.gear_type = gear_type
    
    def drift(self):
        print(f"{self.name} is drifting")
        
    def change_color(self, color):
        self.color = color
        print(f"{self.name} changed the color to {color}")
```

```python
tesla = Car('tesla', 'red', 'electric', True, 'automatic')
tata = Car('tata', 'black', 'petrol', False, 'manual')

# print(tesla.name)
# print(tesla.color)
# print(tesla.engine_type)
# print(tesla.is_self_driving)
# print(tesla.gear_type)

tesla.accelerate()
tesla.brake()

tata.accelerate()
tata.brake()

tesla.drift()
tesla.change_color('white')

# bike_vehicle = Vehicle('honda', 'black', 'petrol')
# print(bike_vehicle.gear_type)
```

```python
class Animal:
    
    def __init__(self, name, type, sound):
        self.name = name
        self.type = type
        self.sound = sound
    
    def walk(self):
        print(f"{self.name} is walking")
    
    def make_sound(self):
        print(f"{self.name} is making a sound like {self.sound}")
        
```

```python
class Pet:
    
    def __init__(self, owner):
        self.owner = owner
    
    def play(self):
        print(f"the pet is playing")
```

```python
class Dog(Animal, Pet):
    def __init__(self, name, type, sound, owner, hobby):
        Animal.__init__(self, name, type, sound)
        Pet.__init__(self, owner)
        self.hobby = hobby
        
    
dog = Dog('shadow', 'dog', 'woof!!', 'shravan', 'playing with bone')

print(dog.name)
print(dog.type)
print(dog.owner)
print(dog.sound)
print(dog.hobby)

dog.walk()
dog.make_sound()
# dog.play()
```
