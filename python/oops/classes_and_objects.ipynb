# Classes and Objects

```python
class Car:
    def __init__(self, name, color, engine_type, speed):
        self.name = name
        self.color = color
        self.engine_type = engine_type
        self.speed = speed
```

```python
car1 = Car('BMW', 'Black', 'Electric', 30)
car2 = Car('KIA', 'White', 'Petrol', 40)

print('properties of BMW car')

print(car1.name)
print(car1.color)
print(car1.engine_type)
print(car1.speed)


print('properties of KIA car')

print(car2.name)
print(car2.color)
print(car2.engine_type)
print(car2.speed)
```

```python
class Car:
    def __init__(self, name, color, engine_type, speed):
        self.name = name
        self.color = color
        self.engine_type = engine_type
        self.speed = speed
        
    
    def accelerate(self):
        print(f"{self.name} is accelearating") 
        
        
    def change_gear(self, gear):
        print(f"{self.name} changes the gear to {gear}")
        
    
    def apply_brake(self):
        print(f"{self.name} has applied brake")
        
```

```python
car1 = Car('BMW', 'Black', 'Electric', 30)

car1.accelerate()
car1.apply_brake()
car1.change_gear(3)


car2 = Car('KIA', 'White', 'Petrol', 40)
car2.accelerate()
car2.apply_brake()
car2.change_gear(2)
```

```python
class Bank:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.balance = 0
        
    def deposit_money(self, money):
        self.balance += money
        print(f"{money} has been added to {self.name} account, current balance is {self.balance}")
    
    def withdraw_money(self, money):
        if self.balance < money:
            print(f"you don't have enough balance")
        self.balance -= money
        print(f"withdrawn {money}, the current balance is {self.balance}")
    
    def get_balance(self):
        print(f"your current balance is {self.balance}")
        


shravan_acc = Bank('shravan', 27)

shravan_acc.get_balance()
shravan_acc.deposit_money(5000)
shravan_acc.withdraw_money(350)
shravan_acc.get_balance()
```
