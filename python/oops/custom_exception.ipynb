```python
class Error(Exception):
    pass

class dobException(Error):
    pass
```

```python
user_age = int(input('Please enter your age'))

try:
    if user_age >= 20 and user_age <= 30:
        print("You can fill this form, you are eligible!")
    else:
        raise dobException
except:
    print("you are not eligible for this exam")
```
