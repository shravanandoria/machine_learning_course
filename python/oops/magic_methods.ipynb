# Magic Methods


Magic methods (also known as dunder methods, short for double underscore) are special, predefined methods in Python that start and end with two underscores.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name}: {self.age}"

    def __repr__(self):
        return "changed the behaviour of repr method"

person = Person('shravan', 27)
# dir(person)

# print(person)
print(repr(person))
```
