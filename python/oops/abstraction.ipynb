# Abstraction 
Abstraction in Python is the process of hiding complex internal implementation details and exposing only the essential features or interfaces to the user. It focuses on what an object does rather than how it does it, reducing code complexity and establishing a clear structure or "contract" for subclasses to follow

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start_engine(self):
        """build own implementation of start_engine method"""
        pass

    @abstractmethod
    def apply_brake(self):
        """build own implementation of brake method"""
        pass
```

```python
class Car(Vehicle):

    def __init__(self):
        pass

    def start_engine(self):
        return f"car engine has started"

    def apply_brake(self):
        return f"car has applied brake"
```

```python
class Bike(Vehicle):
    def __init__(self):
        pass

    def start_engine(self):
        return f"bike engine has started"

    def apply_brake(self):
        return f"bike has applied brake" 
```

```python
car = Car()
bike = Bike()

print(car.start_engine())
print(car.apply_brake())

print(bike.start_engine())
print(bike.apply_brake())
```
