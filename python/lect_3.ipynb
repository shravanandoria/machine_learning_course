```python
# Loop Control Statements
#     . break
#     . continue
#     . pass
```

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
for i in range(10):
    if i == 5:
        continue
    print(i)
```

```python
for i in range(10):
    if i == 5:
        pass
    print(i)
```

```python
age = 15

if age > 18:
    pass
else:
    print("you are under 18")
```

```python
for i in range(1, 6):
    for j in range(1, 6):
        print(i * j, end=" ")
    print()
```

```python
for row in range(5):
    for col in range(row):
        print("*", end=" ")
    print()
        
```

```python
for i in range(20):
    if i % 2 == 0:
        print(i)
```

```python
for i in range(20):
    if i % 2 != 0:
        print(i)
```

```python
for num in range(1, 101):
    if num > 1:
        for i in range(2, num):
            if num % i == 0:
                break
        else:
            print(num)
```
