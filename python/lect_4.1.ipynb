```python
# Shallow Copy

student = {"name": "Shravan", "age": 27, "height": 6.0}

student_copy = student.copy()

print(student)
print(student_copy)
```

```python
student["name"] = "Alex"
print(student)
print(student_copy)
```

```python
student = {"name": "Shravan", "age": 27, "height": 6.0}

for key in student.keys():
    print(key)
```

```python
for value in student.values():
    print(value)
```

```python
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
students = {
    "student1": {
        "name": "shravan",
        "age": 27,
        "grade": "A"
    },
    "student2": {
        "name": "Json",
        "age": 33,
        "grade": "A"
    }
}
```

```python
print(students["student1"]["age"])
```

```python
students.items()
```

```python
for studentId, studentValue in students.items():
    print(studentId, studentValue)
    for key,value in studentValue.items():
        print(f"{key}: {value}")
```

```python
# Dictionary Comprehension

{x:x**2 for x in range(10)}
```

```python

{x:x**2 for x in range(10) if x % 2 == 0}
```

```python
nums = [1,1,1,2,2,2,2,3,4,4,4,4]
freq = {}

for num in nums:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
        
print(freq)
```

```python
dict1 = {'a': 22, 'b': 44}
dict2 = {'c': 98, 'd': 12}

combined_dict = {**dict1, **dict2}
print(combined_dict)
```
