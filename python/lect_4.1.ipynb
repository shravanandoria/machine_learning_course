```python
my_info = {"name": "Shravan", "age": 27, "height": 6.0}
print(type(my_info))
print(my_info)
```

```python
print(my_info["name"])
print(my_info["age"])
print(my_info["height"])
```

```python
print(my_info.get("name"))
print(my_info.get("age"))
print(my_info.get("height"))
```

```python
my_info = {"name": "Shravan", "age": 27, "height": 6.0}

print(my_info)

my_info["name"] = "Robert"
my_info["age"] = 33
my_info["height"] = 5.11
my_info["can_drive"] = True
my_info["is_student"] = True

print(my_info)
```

```python
del my_info["is_student"]
print(my_info)
del my_info["age"]
print(my_info)
```

```python
# Shallow Copy

student = {"name": "Shravan", "age": 27, "height": 6.0}

student_copy = student.copy()

print(student)
print(student_copy)
```

```python
student["name"] = "Alex"
print(student)
print(student_copy)
```
