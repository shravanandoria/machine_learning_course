Importing Modules In Python: Modules & Packages

In Python, modules and packages allow you to break down large programs into small, manageable, and reusable pieces.

```python
import math

print(math.sqrt(16))
```

```python
from math import sqrt, pi


print(math.sqrt(25))
print(pi)
```

```python
import numpy 

arr= numpy.array([1,2,3,4,5])
print(arr)
```

```python
from math import *

print(sqrt(64))
print(pi)
```
