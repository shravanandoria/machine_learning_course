```python
age = 25

if age > 18:
    print("you can drive the car")
else:
    print("you cannot drive the car")
    
```

```python
def check_age_to_drive(age):
    if age > 18:
        print("you can drive the car")
    else:
        print("you cannot drive the car")
    
check_age_to_drive(25)
check_age_to_drive(12)
```

```python
def greet(name="Guest"):
    print(f"Hello {name} how are you?")
    

greet("Alex")
greet()
```

```python
def addition(a,b,c):
    sum = a + b + c
    print(sum)
    
addition(10,2,5)
```

```python
def multiplication(a,b,c,d):
    result = a * b * c * d
    print(result)

multiplication(10,2,3,5)
```

```python
# Variable Length Arguments
# Positional & Keyword Arguments

def print_names(*names):
    for name in names:
        print(name)
        
print_names("tom", "alex", "bob", "mike", 1, True)
```

```python
def print_pairs(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
        
print_pairs(name="Shravan", height=6.0)
```

```python
def addition(a,b,c):
    return a + b + c
    
    
print(addition(2,5,8))
print(addition(7,1,2))
print(addition(1,1,0))
```
