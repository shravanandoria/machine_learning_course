# Exceptions

In Python, exceptions are runtime errors that disrupt the normal flow of a program's execution. Unlike syntax errors (which prevent code from compiling), exceptions occur while the code is actively running.

```python
# ZeroDivisionError - Dividing by zero.

try:
    number = int(input("please enter a number"))
    result = 10 / number    
    print(result)
except ZeroDivisionError as ex:
    print("something went wrong")
    print(ex)
    print("please enter a number greater than 0")
```

```python
# try, except, else

try:
    number = int(input("please enter a number"))
    result = 10 / number    
    print(result)
except Exception as ex:
    print(ex)
else:
    print('the program ran successfully')
```

```python
# try, except, else, finally

try:
    number = int(input("please enter a number"))
    result = 10 / number    
    print(result)
except Exception as ex:
    print(ex)
else:
    print('the program ran successfully')
finally:
    print('finally block ran')

```

```python
# FileNotFoundError - File not found.

try:
    with open('animals.txt', 'r') as file:
        content = file.read()
        print(content)
except FileNotFoundError as ex:
    print("no such file exists")
finally:
    print("we are in finally block")
    
    if 'file' in locals() or not file.closed:
        file.close()
        print('file is closed')


```
