```python
age = 15

if age > 18:
    print("you can legally drive a car")
else:
    print("you cannot legally drive a car")
```

```python
user_age = int(input("put your age to verify if you can drive a car: "))


if user_age > 18:
    print("you can legally drive a car")
else:
    print("you cannot legally drive a car")
```

```python
user_age = int(input("Type your age for discount: "))


if user_age <= 13:
    print("Child discount will be applied, 20%")
elif user_age <= 20:
    print("Teenage discount will be applied, 30%")
elif user_age <= 50:
    print("Adult discount will be applied, 30%")
else:
    print("Senior Citizen discount will be applied, 50%")
    
    
```

```python
age = 30
is_student = True

if age <= 25 and is_student:
    print("You can enter this event")
else:
    print("You are not qualified to enter this event")
```

```python
is_student = False
is_senior_citizen = False

if is_student or is_senior_citizen:
    print("You are allowed to this special event")
else:
    print("You are cannot enter this event")
    
```

```python
age = 20
is_student = False


if age > 18:
    if is_student:
        print("You are above 18 & a student")
    else:
        print("You are above 18 but not a student")
else:
    print("You are not above 18")
```
